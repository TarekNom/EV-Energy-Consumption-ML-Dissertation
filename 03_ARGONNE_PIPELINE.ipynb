# Stage 5 — Safe Loader

```python
# ==========================================================
# Stage 5 - Safe Argonne Loader
# ==========================================================

def load_argonne_file(file_path):
    """
    Load one Argonne BMW test file safely.
    """

    df = pd.read_csv(
        file_path,
        sep="\t",
        encoding="latin1"
    )

    # Remove pre-test measurements
    df = df[
        df["Time[sec]"] >= 0
    ].copy()

    # Add trip identifier
    trip_id = file_path.stem.replace(" Test Data", "")

    df["Trip_ID"] = trip_id

    return df
```

```python
# Test loader

test_trip = load_argonne_file(
    test_files[0]
)

print(test_trip.shape)

display(test_trip.head())

print()

print(test_trip["Trip_ID"].unique())
```

# Stage 6 — Load Test Files

```python
# ==========================================================
# Stage 6 - Load All Test Files
# ==========================================================

all_trips = []

for file_path in test_files:
    try:
        trip = load_argonne_file(file_path)
        all_trips.append(trip)
    except Exception as e:
        print(f"Failed: {file_path.name}")
        print(e)

argonne_all = pd.concat(
    all_trips,
    ignore_index=True
)

print("=" * 60)
print("ARGONNE COMBINED DATASET")
print("=" * 60)

print("\nShape:")
print(argonne_all.shape)

print("\nNumber of trips:")
print(argonne_all["Trip_ID"].nunique())

print("\nTrip IDs:")
print(sorted(argonne_all["Trip_ID"].unique()))
```

# Stage 7 — Temporal and Signal Audit

```python
# ==========================================================
# Stage 7 - Temporal Audit
# ==========================================================

print("=" * 60)
print("TEMPORAL AUDIT")
print("=" * 60)

audit = []

for trip_id, trip in argonne_all.groupby("Trip_ID"):

    delta_t = trip["Time[sec]"].diff().dropna()

    audit.append({
        "Trip_ID": trip_id,
        "Rows": len(trip),
        "Start_Time_s": trip["Time[sec]"].min(),
        "End_Time_s": trip["Time[sec]"].max(),
        "Median_dt_s": delta_t.median(),
        "Maximum_dt_s": delta_t.max(),
        "Missing_TimeSteps": (delta_t > 0.11).sum()
    })

audit_df = pd.DataFrame(audit)

display(audit_df)

print("\nSummary:")
display(audit_df.describe())
```

```python
print("Trips with missing timestamps:")

display(
    audit_df[
        audit_df["Missing_TimeSteps"] > 0
    ]
)
```

# Stage 8 — Energy Target Construction

## Step 8.1 — Battery Power

```python
# ==========================================================
# Stage 8.1 - Battery Power
# ==========================================================

# Battery power (W)
argonne_all["Battery_Power_W"] = (
    argonne_all["HV_Battery_Voltage[V]"] *
    argonne_all["HV_Battery_Current[A]"]
)

print("Battery power statistics:")

display(
    argonne_all["Battery_Power_W"].describe()
)

print("\nCharging samples:")
print((argonne_all["Battery_Power_W"] > 0).sum())

print("\nDischarging samples:")
print((argonne_all["Battery_Power_W"] < 0).sum())

print("\nZero-power samples:")
print((argonne_all["Battery_Power_W"] == 0).sum())
```

## Stage 8.1b

```python
# ==========================================================
# Verify current sign convention
# ==========================================================

print("Battery current statistics:\n")

display(
    argonne_all["HV_Battery_Current[A]"].describe()
)

print("\nPositive current samples:")
print((argonne_all["HV_Battery_Current[A]"] > 0).sum())

print("\nNegative current samples:")
print((argonne_all["HV_Battery_Current[A]"] < 0).sum())

print("\nZero current samples:")
print((argonne_all["HV_Battery_Current[A]"] == 0).sum())
```

## Stage 8.2 — Energy Integration

```python
# ==========================================================
# Stage 8.2 - Energy Integration
# ==========================================================

trip_energy = []

for trip_id, trip in argonne_all.groupby("Trip_ID"):

    trip = trip.sort_values("Time[sec]")

    # Time step (seconds)
    dt = trip["Time[sec]"].diff().fillna(0)

    # Incremental energy (Wh)
    trip["Energy_Increment_Wh"] = (
        trip["Battery_Power_W"] * dt / 3600
    )

    # Separate consumption and regeneration
    discharge_energy = (
        trip.loc[
            trip["Energy_Increment_Wh"] > 0,
            "Energy_Increment_Wh"
        ].sum()
    )

    regen_energy = (
        -trip.loc[
            trip["Energy_Increment_Wh"] < 0,
            "Energy_Increment_Wh"
        ].sum()
    )

    net_energy = discharge_energy - regen_energy

    trip_energy.append({

        "Trip_ID": trip_id,

        "Net_Energy_Wh": net_energy,

        "Discharge_Energy_Wh": discharge_energy,

        "Regenerated_Energy_Wh": regen_energy

    })

trip_energy = pd.DataFrame(trip_energy)

display(trip_energy)

print()

print("Summary statistics:")

display(trip_energy.describe())
```

## Stage 8.3 — Distance Calculation

```python
# ==========================================================
# Stage 8.3 - Distance Calculation
# ==========================================================

# Convert speed from mph to km/h
argonne_all["Speed_kmh"] = (
    argonne_all["Dyno_Speed[mph]"] * 1.60934
)

trip_distance = []

for trip_id, trip in argonne_all.groupby("Trip_ID"):

    trip = trip.sort_values("Time[sec]")

    dt_hours = trip["Time[sec]"].diff().fillna(0) / 3600

    distance = (
        trip["Speed_kmh"] * dt_hours
    ).sum()

    trip_distance.append({

        "Trip_ID": trip_id,

        "Calculated_Distance_km": distance

    })

trip_distance = pd.DataFrame(trip_distance)

display(trip_distance)

print()

print("Distance statistics:")

display(
    trip_distance["Calculated_Distance_km"].describe()
)
```

## Stage 8.4 — Build the Target


```python
# ==========================================================
# Stage 8.4 - Energy Consumption Target
# ==========================================================

# Merge energy and distance
trip_summary = trip_energy.merge(
    trip_distance,
    on="Trip_ID"
)

# Target (kWh/100 km)
trip_summary["Energy_Consumption_kWh_100km"] = (
    trip_summary["Net_Energy_Wh"] /
    trip_summary["Calculated_Distance_km"]
) / 10

print("=" * 60)
print("ARGONNE TARGET")
print("=" * 60)

display(trip_summary)

print("\nTarget statistics:")

display(
    trip_summary["Energy_Consumption_kWh_100km"].describe()
)
```

# Stage 9 — Feature Engineering


## Step 9.1 — Calculate Acceleration

```python
# ==========================================================
# Stage 9.1 - Calculate Speed and Acceleration
# ==========================================================

# Speed already converted previously
# Calculate longitudinal acceleration

argonne_all = argonne_all.sort_values(
    ["Trip_ID", "Time[sec]"]
).reset_index(drop=True)

argonne_all["Acceleration_mps2"] = (
    argonne_all
        .groupby("Trip_ID")["Speed_kmh"]
        .diff()
        / 3.6
) / (
    argonne_all
        .groupby("Trip_ID")["Time[sec]"]
        .diff()
)

argonne_all["Acceleration_mps2"] = (
    argonne_all["Acceleration_mps2"]
    .fillna(0)
)

print("Acceleration statistics:")

display(
    argonne_all["Acceleration_mps2"].describe()
)
```

## Step 9.2 — Build the Machine Learning Features


```python
# ==========================================================
# Stage 9.2 - Trip-level Feature Engineering
# ==========================================================

feature_table = []

for trip_id, trip in argonne_all.groupby("Trip_ID"):

    duration = (
        trip["Time[sec]"].max() -
        trip["Time[sec]"].min()
    )

    feature_table.append({

        "Trip_ID": trip_id,

        # Driving
        "Mean_Speed_kmh": trip["Speed_kmh"].mean(),
        "Max_Speed_kmh": trip["Speed_kmh"].max(),
        "Std_Speed_kmh": trip["Speed_kmh"].std(),

        "Mean_Acceleration_mps2": trip["Acceleration_mps2"].mean(),
        "Max_Acceleration_mps2": trip["Acceleration_mps2"].max(),
        "Min_Acceleration_mps2": trip["Acceleration_mps2"].min(),

        "Mean_Positive_Acceleration_mps2":
            trip.loc[
                trip["Acceleration_mps2"] > 0,
                "Acceleration_mps2"
            ].mean(),

        "Idling_Proportion":
            (trip["Speed_kmh"] < 1).mean(),

        # Battery
        "Start_SOC":
            trip["HV_Battery_SOC[%]"].iloc[0],

        "End_SOC":
            trip["HV_Battery_SOC[%]"].iloc[-1],

        # Environment
        "Mean_Test_Cell_Temp_C":
            trip["Test_Cell_Temp[C]"].mean(),

        "Mean_Test_Cell_RH":
            trip["Test_Cell_RH[%]"].mean(),

        # Vehicle
        "Mean_Motor_RPM":
            trip["Motor_Speed[rpm]"].mean(),

        "Mean_Tractive_Effort_N":
            trip["Dyno_Tractive_Effort[N]"].mean(),

        "Mean_Accelerator_Position":
            trip["Accelerator_Pedal_Position[%]"].mean(),

        "Duration_s":
            duration

    })

feature_table = pd.DataFrame(feature_table)

print(feature_table.shape)

display(feature_table.head())
```

## Stage 9.3 — Merge the Target

```python
# ==========================================================
# Stage 9.3 - Merge Target
# ==========================================================

argonne_ml = feature_table.merge(
    trip_summary[
        [
            "Trip_ID",
            "Calculated_Distance_km",
            "Net_Energy_Wh",
            "Energy_Consumption_kWh_100km"
        ]
    ],
    on="Trip_ID",
    how="left"
)

print("=" * 60)
print("ARGONNE ML TABLE")
print("=" * 60)

print("\nShape:")
print(argonne_ml.shape)

print("\nMissing values:")
display(argonne_ml.isna().sum())

print("\nDuplicate trips:")
print(argonne_ml["Trip_ID"].duplicated().sum())

display(argonne_ml.head())
```

## Stage 9.5 — Target Validation

```python
# ==========================================================
# Stage 9.5 - Target Validation
# ==========================================================

print("=" * 60)
print("TARGET VALIDATION")
print("=" * 60)

print("\nSummary statistics:")

display(
    argonne_ml["Energy_Consumption_kWh_100km"].describe()
)

print("\nNegative consumption trips:")

display(
    argonne_ml[
        argonne_ml["Energy_Consumption_kWh_100km"] < 0
    ][
        [
            "Trip_ID",
            "Calculated_Distance_km",
            "Net_Energy_Wh",
            "Energy_Consumption_kWh_100km"
        ]
    ]
)

print("\nTrips below 5 kWh/100km:")

display(
    argonne_ml[
        (argonne_ml["Energy_Consumption_kWh_100km"] >= 0) &
        (argonne_ml["Energy_Consumption_kWh_100km"] < 5)
    ][
        [
            "Trip_ID",
            "Energy_Consumption_kWh_100km"
        ]
    ]
)

print("\nTrips above 40 kWh/100km:")

display(
    argonne_ml[
        argonne_ml["Energy_Consumption_kWh_100km"] > 40
    ][
        [
            "Trip_ID",
            "Energy_Consumption_kWh_100km"
        ]
    ]
)
```

# Stage 10 — Leakage-Safe Modelling Table

```python
# ==========================================================
# Stage 10 - Leakage-Safe ML Table
# ==========================================================

# Keep a complete audit table
argonne_audit = argonne_ml.copy()

# Remove regeneration-dominated trips
argonne_ml_final = argonne_ml[
    argonne_ml["Energy_Consumption_kWh_100km"] >= 0
].copy()

# Remove leakage columns
argonne_ml_final = argonne_ml_final.drop(
    columns=[
        "End_SOC",
        "Calculated_Distance_km",
        "Net_Energy_Wh"
    ]
)

print("=" * 60)
print("ARGONNE LEAKAGE-SAFE ML TABLE")
print("=" * 60)

print("\nAudit table shape:")
print(argonne_audit.shape)

print("\nML table shape:")
print(argonne_ml_final.shape)

print("\nColumns:")

for c in argonne_ml_final.columns:
    print("-", c)

print("\nMissing values:")

display(
    argonne_ml_final.isna().sum()
)

display(argonne_ml_final.head())
```
